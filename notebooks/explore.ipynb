# Выручка за 23 августа 2026

Смотрю, сколько ПРАЙМ заработал за день, и сравниваю с неделей раньше.

```python
import pandas as pd

payments = pd.read_csv("../data/payments.csv", parse_dates=["paid_at"])
payments.shape
```

```python
day = "2026-08-23"
```

```python
paid = payments[payments["status"] == "success"]
```

```python
day_revenue = paid.loc[paid["paid_at"].dt.strftime("%Y-%m-%d") == day, "amount"].sum()
day_revenue
```
